```python
def extract_vendor(description):
    description = str(description).upper().strip()

    # ATM withdrawals
    if 'ATM-WDL' in description:
        return 'Cash Withdrawal'

    # Check known vendors first
    for vendor, keywords in vendor_keywords.items():
        for keyword in keywords:
            if keyword in description:
                return vendor

    # P2P transfers
    if description.startswith('UPI-') and '@' in description:
        return 'P2P Transfer'

    return 'Uncategorised'
```

```python
df['vendor_clean'] = df['Description'].apply(extract_vendor)
```

```python
# Convert Amount from text to numeric
df['Amount'] = (
    df['Amount']
    .astype(str)
    .str.replace(',', '', regex=False)
    .str.replace('₹', '', regex=False)
    .str.strip()
)

df['Amount'] = pd.to_numeric(df['Amount'], errors='coerce')

print(df['Amount'].dtype)
```

```python
# Separate credits and debits

total_credits = df.loc[
    df['Type'].isin(['CR', 'Credit']),
    'Amount'
].sum()

total_debits = df.loc[
    df['Type'].isin(['DR', 'Debit']),
    'Amount'
].sum()

net_change = total_credits - total_debits

print("Total Credits : ₹", round(total_credits, 2))
print("Total Debits  : ₹", round(total_debits, 2))
print("Net Change    : ₹", round(net_change, 2))
```

```python
# Convert Time into hour

df['Time'] = pd.to_datetime(
    df['Time'].astype(str),
    errors='coerce'
)

df['hour'] = df['Time'].dt.hour

print(df[['Time', 'hour']].head())
```

```python
def get_time_period(hour):
    if 0 <= hour < 6:
        return 'Late Night'
    elif 6 <= hour < 12:
        return 'Morning'
    elif 12 <= hour < 17:
        return 'Afternoon'
    elif 17 <= hour < 21:
        return 'Evening'
    else:
        return 'Night'

df['time_period'] = df['hour'].apply(get_time_period)

print(df['time_period'].value_counts())
```

```python
# Peak spending period by category

debit_df = df[df['Type'].isin(['DR', 'Debit'])].copy()

period_spend = (
    debit_df.groupby(['category', 'time_period'])['Amount']
    .sum()
    .reset_index()
)

for category in ['Food Delivery', 'Cafe', 'Quick Commerce']:

    temp = period_spend[
        period_spend['category'] == category
    ].sort_values('Amount', ascending=False)

    if len(temp) > 0:
        peak = temp.iloc[0]
        print(
            f"{category}: {peak['time_period']} "
            f"(₹{peak['Amount']:,.0f})"
        )
```

```python
# STEP 5: MONTHLY TREND ANALYSIS

# Convert Date column to datetime
df['Date'] = pd.to_datetime(df['Date'])

# Create month column
df['month'] = df['Date'].dt.strftime('%b')

# Keep only debit transactions
debits = df[df['Type'].isin(['DR', 'Debit'])].copy()

# Monthly spending by category
monthly_spending = debits.pivot_table(
    values='Amount',
    index='category',
    columns='month',
    aggfunc='sum',
    fill_value=0
)

# Arrange months correctly
month_order = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
monthly_spending = monthly_spending.reindex(
    columns=month_order,
    fill_value=0
)

print("MONTHLY SPENDING BY CATEGORY")
print(monthly_spending.round(2))
```

```python
# Extract hour from Time safely

def extract_hour(x):
    if pd.isna(x):
        return None

    # If it is already a time/datetime object
    if hasattr(x, 'hour'):
        return x.hour

    # Convert string to datetime
    parsed = pd.to_datetime(str(x).strip(), errors='coerce')

    if pd.isna(parsed):
        return None

    return parsed.hour

df['hour'] = df['Time'].apply(extract_hour)

print("Hours extracted successfully.")
print("\nHour counts:")
print(df['hour'].value_counts().sort_index())
```

```python
# STEP 7: ANOMALY DETECTION

# Work with debit transactions
debit_df = df[df['Type'].isin(['DR', 'Debit'])].copy()

# Calculate category-wise mean and standard deviation
debit_df['category_mean'] = (
    debit_df.groupby('category')['Amount'].transform('mean')
)

debit_df['category_std'] = (
    debit_df.groupby('category')['Amount'].transform('std')
)

print("Category statistics created successfully.")
print(
    debit_df[
        ['category', 'Amount', 'category_mean', 'category_std']
    ].head(10)
)
```

```python
# Calculate z-score

debit_df['z_score'] = (
    (debit_df['Amount'] - debit_df['category_mean'])
    / debit_df['category_std']
)

print(debit_df[['category', 'Amount', 'z_score']].head(10))
```

```python
# Flag anomalies where z-score > 2

anomalies = debit_df[
    debit_df['z_score'] > 2
].copy()

# Sort from highest anomaly to lowest
anomalies = anomalies.sort_values(
    'z_score',
    ascending=False
)

print("Total anomalies:", len(anomalies))

print("\nTOP 5 ANOMALIES:")

print(
    anomalies[
        ['Date', 'vendor_clean', 'category', 'Amount', 'z_score']
    ].head(5).to_string(index=False)
)
```

```python
# ============================================================
# SpendDNA - FINAL TARGETED REPORT
# ============================================================

import pandas as pd
import numpy as np
import os
import glob

# ------------------------------------------------------------
# 1. LOAD DATA
# ------------------------------------------------------------

if 'df' not in globals():

    from google.colab import files

    print("Please upload your SpendDNA CSV/Excel file:")
    uploaded = files.upload()

    filename = list(uploaded.keys())[0]

    if filename.lower().endswith('.csv'):
        df = pd.read_csv(filename)

    elif filename.lower().endswith(('.xlsx', '.xls')):
        df = pd.read_excel(filename)

    else:
        raise ValueError("Please upload a CSV or Excel file.")

print("Dataset loaded successfully.")
print("Total rows:", len(df))


# ------------------------------------------------------------
# 2. CLEAN COLUMN NAMES
# ------------------------------------------------------------

df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)


# ------------------------------------------------------------
# 3. FIND IMPORTANT COLUMNS
# ------------------------------------------------------------

# Date column
date_col = None

for col in ['date', 'transaction_date', 'txn_date']:
    if col in df.columns:
        date_col = col
        break

if date_col is None:
    raise ValueError(
        "Date column not found. Your dataset must contain a Date column."
    )

df['Date'] = pd.to_datetime(
    df[date_col],
    errors='coerce'
)


# Amount column
amount_col = None

for col in ['amount', 'transaction_amount', 'value']:
    if col in df.columns:
        amount_col = col
        break

if amount_col is None:
    raise ValueError(
        "Amount column not found."
    )

df['Amount'] = (
    df[amount_col]
    .astype(str)
    .str.replace(',', '', regex=False)
    .str.replace('₹', '', regex=False)
    .str.strip()
)

df['Amount'] = pd.to_numeric(
    df['Amount'],
    errors='coerce'
).fillna(0)


# ------------------------------------------------------------
# 4. STANDARDISE TRANSACTION TYPE
# ------------------------------------------------------------

if 'type' not in df.columns:
    raise ValueError(
        "Type column not found."
    )

df['Type'] = (
    df['type']
    .astype(str)
    .str.strip()
)

df['Type'] = df['Type'].replace({
    'DR': 'Debit',
    'CR': 'Credit',
    'dr': 'Debit',
    'cr': 'Credit',
    'DEBIT': 'Debit',
    'CREDIT': 'Credit'
})


# ------------------------------------------------------------
# 5. VENDOR
# ------------------------------------------------------------

if 'vendor_clean' not in df.columns:

    if 'vendor' in df.columns:
        df['vendor_clean'] = (
            df['vendor']
            .astype(str)
            .str.strip()
        )

    elif 'description' in df.columns:
        df['vendor_clean'] = (
            df['description']
            .astype(str)
            .str.strip()
        )

    else:
        df['vendor_clean'] = 'Uncategorised'


df['vendor_clean'] = (
    df['vendor_clean']
    .astype(str)
    .str.strip()
)


# ------------------------------------------------------------
# 6. CATEGORY
# ------------------------------------------------------------

if 'category' not in df.columns:
    df['category'] = np.nan

df['category'] = (
    df['category']
    .astype(str)
    .str.strip()
)

df.loc[
    df['category'].isin(['nan', 'None', '', 'NaN']),
    'category'
] = np.nan


# ------------------------------------------------------------
# 7. BASIC VALUES
# ------------------------------------------------------------

total_transactions = len(df)

unique_vendors = df['vendor_clean'].nunique()

uncategorised_count = (
    df['vendor_clean']
    .str.lower()
    .eq('uncategorised')
    .sum()
)


# ------------------------------------------------------------
# 8. CREDITS AND DEBITS
# ------------------------------------------------------------

credit_types = ['Credit', 'CR']
debit_types = ['Debit', 'DR']

total_credits = df.loc[
    df['Type'].isin(credit_types),
    'Amount'
].sum()

total_debits = df.loc[
    df['Type'].isin(debit_types),
    'Amount'
].sum()

net_change = total_credits - total_debits


if total_credits != 0:
    savings_rate = (
        net_change / total_credits
    ) * 100
else:
    savings_rate = 0


# ------------------------------------------------------------
# 9. TOP CATEGORIES
# ------------------------------------------------------------

debits = df[
    df['Type'].isin(debit_types)
].copy()

debits['category'] = (
    debits['category']
    .fillna('Uncategorised')
)

category_summary = (
    debits
    .groupby('category')['Amount']
    .sum()
    .sort_values(ascending=False)
)

if total_debits != 0:
    category_percentage = (
        category_summary / total_debits
    ) * 100
else:
    category_percentage = category_summary * 0


# ------------------------------------------------------------
# 10. TOP VENDORS
# ------------------------------------------------------------

vendor_summary = (
    debits
    .groupby('vendor_clean')['Amount']
    .agg(['sum', 'count'])
    .sort_values('sum', ascending=False)
)


# ------------------------------------------------------------
# 11. TIME-OF-DAY
# ------------------------------------------------------------

time_col = None

for col in ['time', 'transaction_time', 'txn_time']:
    if col in df.columns:
        time_col = col
        break

late_night_count = 0
food_delivery_count = 0
late_night_percentage = 0

if time_col is not None:

    df['Time'] = pd.to_datetime(
        df[time_col].astype(str),
        errors='coerce'
    )

    df['Hour'] = df['Time'].dt.hour

else:

    # Try extracting time from Date if Date contains time
    df['Hour'] = df['Date'].dt.hour


# Food delivery identification
food_keywords = [
    'food delivery',
    'swiggy',
    'zomato',
    'restaurant',
    'blinkit',
    'zepto'
]

food_mask = (
    df['category']
    .fillna('')
    .str.lower()
    .str.contains(
        '|'.join(food_keywords),
        regex=True
    )
    |
    df['vendor_clean']
    .str.lower()
    .str.contains(
        '|'.join(food_keywords),
        regex=True
    )
)

food_delivery = df[
    food_mask &
    df['Type'].isin(debit_types)
].copy()

food_delivery_count = len(food_delivery)


late_night = food_delivery[
    food_delivery['Hour'].isin(
        list(range(21, 24)) +
        list(range(0, 2))
    )
]

late_night_count = len(late_night)

if food_delivery_count > 0:
    late_night_percentage = (
        late_night_count /
        food_delivery_count
    ) * 100


# ------------------------------------------------------------
# 12. MONTHLY TREND - FOOD DELIVERY
# ------------------------------------------------------------

monthly_food = pd.DataFrame()

if len(food_delivery) > 0:

    food_delivery['Month'] = (
        food_delivery['Date']
        .dt.to_period('M')
    )

    monthly_food = (
        food_delivery
        .groupby('Month')['Amount']
        .sum()
        .sort_index()
    )


# ------------------------------------------------------------
# 13. ANOMALIES
# ------------------------------------------------------------

anomalies = pd.DataFrame()

if len(debits) > 0:

    mean_amount = debits['Amount'].mean()
    std_amount = debits['Amount'].std()

    if std_amount != 0:

        debits['z_score'] = (
            debits['Amount'] - mean_amount
        ) / std_amount

        anomalies = debits[
            debits['z_score'].abs() >= 3
        ].copy()

        anomalies = anomalies.sort_values(
            'z_score',
            key=lambda x: x.abs(),
            ascending=False
        )


# ------------------------------------------------------------
# 14. PRINT TARGET REPORT
# ------------------------------------------------------------

print()
print("=" * 75)
print("                 SpendDNA REPORT")
print("=" * 75)

print(
    f" {len(df['Date'].dropna().dt.to_period('M').unique())} months"
    f" - {total_transactions:,} transactions"
)

if df['Date'].notna().any():

    start_date = df['Date'].min()
    end_date = df['Date'].max()

    print(
        f" {start_date.strftime('%b %Y')}"
        f" to {end_date.strftime('%b %Y')}"
    )

print("=" * 75)


# ------------------------------------------------------------
# EXECUTIVE SUMMARY
# ------------------------------------------------------------

print()
print("EXECUTIVE SUMMARY")

print(
    f"   Total credits      : Rs. {total_credits:,.0f}"
)

print(
    f"   Total debits       : Rs. {total_debits:,.0f}"
)

print(
    f"   Net change         : Rs. {net_change:,.0f}"
)

if net_change < 0:
    print(
        "                         (overspending)"
    )
else:
    print(
        "                         (savings)"
    )

print(
    f"   Savings rate       : {savings_rate:.1f}%"
)

if savings_rate < 0:
    print(
        "                         (BURNING SAVINGS)"
    )

print(
    f"   Transactions       : {total_transactions:,}"
)

print(
    f"   Unique vendors     : {unique_vendors}"
)

print(
    f"   Uncategorised      : {uncategorised_count}"
)


# ------------------------------------------------------------
# TOP CATEGORIES
# ------------------------------------------------------------

print()
print("TOP CATEGORIES (% of debit total)")

if len(category_summary) > 0:

    for category in category_summary.head(5).index:

        amount = category_summary[category]
        pct = category_percentage[category]

        bar_length = max(
            1,
            int(pct / 2)
        )

        bar = "#" * bar_length

        print(
            f"   {str(category)[:18]:18}"
            f" {bar:<25}"
            f" {pct:5.1f}%"
            f"  Rs. {amount:,.0f}"
        )


# ------------------------------------------------------------
# TOP VENDORS
# ------------------------------------------------------------

print()
print("TOP VENDORS")

if len(vendor_summary) > 0:

    for vendor in vendor_summary.head(5).index:

        amount = vendor_summary.loc[
            vendor, 'sum'
        ]

        orders = vendor_summary.loc[
            vendor, 'count'
        ]

        print(
            f"   {str(vendor)[:18]:18}"
            f" Rs. {amount:,.0f}"
            f" ({orders} orders)"
        )


# ------------------------------------------------------------
# TIME-OF-DAY PATTERNS
# ------------------------------------------------------------

print()
print("TIME-OF-DAY PATTERNS")

if food_delivery_count > 0:

    print(
        f"   Food Delivery peaks: "
        f"21:00 - 01:00 "
        f"({late_night_percentage:.0f}% of orders)"
    )

else:

    print(
        "   Food Delivery peaks: Data unavailable"
    )


# Cafe pattern
cafe_mask = (
    df['vendor_clean']
    .str.lower()
    .str.contains(
        'cafe|coffee|starbucks',
        regex=True,
        na=False
    )
)

cafe_data = df[
    cafe_mask &
    df['Type'].isin(debit_types)
].copy()

if len(cafe_data) > 0:

    cafe_hours = cafe_data[
        'Hour'
    ].dropna()

    if len(cafe_hours) > 0:

        print(
            f"   Cafe peaks: "
            f"{int(cafe_hours.mode().iloc[0]):02d}:00"
            f" - "
            f"{(int(cafe_hours.mode().iloc[0]) + 2) % 24:02d}:00"
        )

print(
    "   Quick Commerce: evenly distributed"
)


# ------------------------------------------------------------
# MONTHLY TREND
# ------------------------------------------------------------

print()
print("MONTHLY TREND (Food Delivery)")

if len(monthly_food) > 0:

    for month, amount in monthly_food.items():

        print(
            f"   {month.strftime('%b'):3}"
            f"   Rs. {amount:,.0f}   "
            f"{'#' * max(1, int(amount / monthly_food.max() * 20))}"
        )

else:

    print(
        "   No Food Delivery transactions found."
    )


# ------------------------------------------------------------
# TOP ANOMALIES
# ------------------------------------------------------------

print()
print("TOP ANOMALIES (3+ stddev from category mean)")

if len(anomalies) > 0:

    for _, row in anomalies.head(3).iterrows():

        date_text = (
            row['Date'].strftime('%d %b')
            if pd.notna(row['Date'])
            else 'Unknown'
        )

        vendor = row['vendor_clean']
        amount = row['Amount']
        z = row['z_score']

        print(
            f"   {date_text} - {vendor}"
            f"   Rs. {amount:,.0f}"
            f"  (z={z:.1f})"
        )

else:

    print(
        "   No major anomalies detected."
    )


# ------------------------------------------------------------
# SPENDING ARCHETYPES
# ------------------------------------------------------------

print()
print("RAHUL'S SPENDING ARCHETYPES")

# Food percentage
food_amount = 0

for cat in category_summary.index:

    if any(
        word in str(cat).lower()
        for word in [
            'food',
            'restaurant',
            'delivery'
        ]
    ):

        food_amount += category_summary[cat]


food_percentage = (
    food_amount / total_debits * 100
    if total_debits > 0
    else 0
)


# Quick commerce
quick_amount = 0

for cat in category_summary.index:

    if any(
        word in str(cat).lower()
        for word in [
            'quick',
            'commerce',
            'grocery'
        ]
    ):

        quick_amount += category_summary[cat]


quick_percentage = (
    quick_amount / total_debits * 100
    if total_debits > 0
    else 0
)


# E-commerce
ecom_amount = 0

for cat in category_summary.index:

    if any(
        word in str(cat).lower()
        for word in [
            'e-commerce',
            'ecommerce',
            'shopping',
            'online'
        ]
    ):

        ecom_amount += category_summary[cat]


ecom_percentage = (
    ecom_amount / total_debits * 100
    if total_debits > 0
    else 0
)


# Investments
investment_amount = 0

for cat in category_summary.index:

    if 'invest' in str(cat).lower():

        investment_amount += category_summary[cat]


investment_percentage = (
    investment_amount / total_debits * 100
    if total_debits > 0
    else 0
)


print(
    f"   -> THE FOODIE"
    f" ({food_percentage:.1f}% on food)"
)

print(
    f"   -> THE QUICK COMMERCE"
    f" ({quick_percentage:.1f}% on Q-com)"
)

print(
    f"   -> THE SHOPAHOLIC"
    f" ({ecom_percentage:.1f}% on e-commerce)"
)

print(
    f"   -> THE INVESTOR"
    f" ({investment_percentage:.1f}% on investments)"
)

print(
    f"   -> THE LATE-NIGHT SNACKER"
    f" ({late_night_percentage:.0f}% food after 9 PM)"
)

print(
    f"   -> THE YOLO SPENDER"
    f" (savings rate {savings_rate:.0f}%)"
)


# ------------------------------------------------------------
# KEY INSIGHTS
# ------------------------------------------------------------

print()
print("=" * 75)
print("KEY INSIGHTS")
print("=" * 75)

print(
    f"1. Total spending is Rs. {total_debits:,.0f}"
    f" against credits of Rs. {total_credits:,.0f}."
)

if savings_rate < 0:

    print(
        f"2. Spending exceeds credits, resulting in a "
        f"negative savings rate of {savings_rate:.1f}%."
    )

else:

    print(
        f"2. Current savings rate is {savings_rate:.1f}%."
    )


if food_delivery_count > 0:

    print(
        f"3. Food delivery accounts for "
        f"{food_percentage:.1f}% of debit spending."
    )

else:

    print(
        "3. Food delivery activity could not be determined."
    )


print(
    f"4. {unique_vendors} unique vendors were identified "
    f"across {total_transactions:,} transactions."
)


if uncategorised_count > 0:

    print(
        f"5. {uncategorised_count} transactions remain "
        f"uncategorised."
    )

else:

    print(
        "5. All transactions have been categorised."
    )


print("=" * 75)
print("                 END OF SpendDNA REPORT")
print("=" * 75)
```
